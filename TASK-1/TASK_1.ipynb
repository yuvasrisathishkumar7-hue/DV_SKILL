###**STEP 1: IMPORT REQUIRED LIBRARIES**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### **STEP 2: LOAD THE SUPERSTORE DATASET**

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/samplesuperstore.csv")
```

```python
df.describe()
```

### **STEP 6: CONVERT ORDER DATE AND SHIP DATE INTO DATETIME FORMAT**

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

### **STEP 7: VERIFY THE UPDATED DATA TYPES**

```python
df.info()
```

### **STEP 8: CALCULATE DELIVERY DAYS**

```python
df['Delivery Days'] = (df['Ship Date'] - df['Order Date']).dt.days
```

### **STEP 9: DISPLAY THE UPDATED DATASET**


```python
df.head()
```

### **STEP 10: DISPLAY UNIQUE VALUES IN THE CATEGORY COLUMN**

```python
df['Category'].unique()
```

### **STEP 11: CHECK FOR MISSING VALUES**

```python
df.isnull().sum()
```

### **STEP 12: CALCULATE TOTAL SALES BY CATEGORY**

```python
category_sales = (df.groupby('Category')['Sales'].sum())
category_sales
```

### **STEP 13: VISUALIZE TOTAL SALES BY CATEGORY**

```python
category_sales.plot( kind='bar',figsize=(8,5),color='#C3B1E1')
plt.title("Sales by Category")
plt.ylabel("Total Sales")
plt.show()
```

### **STEP 14: VISUALIZE THE DISTRIBUTION OF SALES**

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Sales'],bins=30,color='#C3B1E1')
plt.title("Sales Distribution")
plt.show()
```
