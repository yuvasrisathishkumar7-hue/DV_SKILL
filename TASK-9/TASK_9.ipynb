**SELECT PROPERTY FEATURES**

```python
features = [
    'bedrooms',
    'bathrooms',
    'sqft_living',
    'grade',
    'yr_built'
]

print("Property Features:")
print(df[features].head())
```

**EXTRACT SALE YEAR**

```python
df['sale_year'] = pd.to_datetime(
    df['date']
).dt.year
```

**CALCULATE PROPERTY AGE**

```python
df['property_age'] = (
    df['sale_year'] - df['yr_built']
)
```

**CHECK RENOVATION STATUS**

```python
df['is_renovated'] = np.where(
    df['yr_renovated'] > 0,
    'Yes',
    'No'
)

df[
    ['sale_year', 'is_renovated']
].head()
```

**CALCULATE PRICE PER SQUARE FOOT**

```python
df['price_per_sqft'] = (
    df['price'] / df['sqft_living']
)

df[
    ['price', 'sqft_living', 'price_per_sqft']
].head()
```

**CONVERT DATE COLUMN**

```python
df['date'] = pd.to_datetime(
    df['date'],
    errors='coerce'
)
```

**EXTRACT SALE YEAR**

```python
df['sale_year'] = df['date'].dt.year
```

**CALCULATE AGE AT SALE**

```python
df['age_at_sale'] = (
    df['sale_year'] - df['yr_built']
)
```

**CALCULATE YEARS SINCE RENOVATION**

```python
df['years_since_renovation'] = np.where(
    df['yr_renovated'].notna() &
    (df['yr_renovated'] > 0),
    df['sale_year'] - df['yr_renovated'],
    0
)
```

**DISPLAY DATE AND AGE DETAILS**

**CALCULATE PRICE PER SQUARE FOOT OF LIVING AREA**

**CALCULATE PRICE PER SQUARE FOOT OF LOT AREA**

**CALCULATE QUARTILES AND IQR**

```python
Q1 = df.quantile(0.25, numeric_only=True)

Q3 = df.quantile(0.75, numeric_only=True)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR
```

**IDENTIFY OUTLIERS**

```python
df_numeric = df.select_dtypes(
    include=[np.number]
)

outliers = df_numeric[
    (df_numeric < lower_bound) |
    (df_numeric > upper_bound)
]
```

**DISPLAY OUTLIER ANALYSIS**

```python
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print(
    "Number of Outliers:",
    len(outliers.dropna(how='all'))
)
```
